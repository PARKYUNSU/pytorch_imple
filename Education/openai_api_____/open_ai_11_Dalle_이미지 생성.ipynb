```python
client = OpenAI()
```

```python
response =\
client.images.generate(
    model='dall-e-3',
    prompt='하얀 샴 고양이',
    size = '1024x1024',    # 이미지 크기
    quality = 'hd',  # 품질
    response_format = 'b64_json',
    n=1                    # 생성할 이미지 개수
)
```

```python
import os
import io
import base64

from PIL import Image
# PIL (python imaging library)
```

```python
image_data = base64.b64decode(response.data[0].b64_json)
# base64.b64decode : Base64로 쓰인 데이터 >> 원복하기 위해 (원래 이미지 파일형태로 반환하기 위함)
```

```python
image = Image.open(io.BytesIO(image_data))
image

# Image.open() : 파일처럼 만들어진 이미지 데이터를 실제 열어서 볼 수 있게 해주는 명령
# io.BytesIO() : image_data를 BytesIO 객체 만들어 줌 >> io 에서 메모리에서 바이트 데이터를 다룰 수 있는 파일 변환
```
