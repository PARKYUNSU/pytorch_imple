```python
speech_file_path = '/content/drive/MyDrive/speech.mp3'
```

```python
client = OpenAI()
```

```python
response = client.audio.speech.create(
    model = 'tts-1',
    voice = 'alloy',
    input = 'Today is a beautiful day to date with someone who I love'
)

response.stream_to_file(speech_file_path)
```
