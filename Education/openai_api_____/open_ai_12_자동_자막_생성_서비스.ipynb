```python
# 음성파일(mp3) 경로가 들어오면 자막 파일(srt) 형식 의 텍스트를 반환
#https://www.youtube.com/shorts/0xvNed3G-18

def get_transcribe(file_path):
  audio_file = open(file_path, 'rb')
  transcript = client.audio.transcriptions.create(
                model = 'whisper-1',
                response_format = 'srt',
                file = audio_file
            )
  return transcript
```

```python
# 유튜브 주소를 입력하면 영상(mp4) 파일과 음성 파일(mp3) 다운로드
def get_video_and_audio(url):
  if url != '':
    yt = YouTube(url, on_progress_callback=on_progress)

    # 오디오 스트림 선택 >> 다운로드(mp3)
    #audio = yt.streams.filter(only_audio=True).first()
    # 유튜브 영상에서 오디오만 추출
    ys = yt.streams.get_audio_only()
    new_audio_file = ys.download(mp3=True)

    # 비디오 스트림 선택 >> 다운로드(mp4)
    # 유튜브 영상 중 mp4 형식의 비디오 스트림 선택
    video = yt.streams.get_highest_resolution()
    video_file = ys.download()

    # 파일크기 로깅(logging: 로그에 파일 크기를 기록함)
    audio_file_stats = os.stat(new_audio_file) # stat(status)
    video_file_stats = os.stat(video_file)
    logging.info(f'audio file size: {audio_file_stats.st_size}')
    logging.info(f'video file size: {video_file_stats.st_size}')

    return new_audio_file, video_file
```

```python
# 유튜브 링크 일력으로 하여 영상 파일, 음성 파일 현재 경로에 저장
url = 'https://www.youtube.com/shorts/0xvNed3G-18'
audio_file, video_file = get_video_and_audio(url)
```
