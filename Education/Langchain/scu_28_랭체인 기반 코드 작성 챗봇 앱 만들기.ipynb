````python
from langchain.chat_models import ChatOpenAI
from langchain.prompts import ChatPromptTemplate, SystemMessagePromptTemplate, HumanMessagePromptTemplate
from langchain.schema.output_parser import StrOutputParser
from langchain_experimental.utilities import PythonREPL
# langchain env 에서  PythonREPL(read, eval-print loop)

template = """Write some python code to solve the user's problem.
Do not ask the user again for confirmation or additional information.

Return only python code in Markdown format, e.g.:

```python
....
```"""

chat_prompt_template = ChatPromptTemplate.from_messages(
              [("system", template),
              ("human","{user_input}")])

# 파이썬 코드만 추출

'''
```python
if x > 3:
    print("A")
else
    print("B")
```

_sanitize_output을 이용하여, 위 출력을 아래처럼 파이썬 코드만 추출한다.

if x > 3:
    print("A")
else
    print("B")

'''
# 주어진 문자열(text)에서 특정 부분만 추출
def _sanitize_output(text: str):
  print(text)

  _, after = text.split("```python")
  # _ : 첫번째 요소는 무시, 두번째 요소 >> after 변수

  return after.split("```")[0] # code 만 가져

chain =\
(chat_prompt_template
 | ChatOpenAI()
 | StrOutputParser()
 | _sanitize_output
 | PythonREPL().run)

def llm(input_text):
    output = chain.invoke({"user_input": input_text})
    return output
````

````python
output = llm("구구단을 실행해줘")
````

````python
print(output)
````

````python
def chat_with_user(user_message):
  ai_message = llm(user_message)
  return ai_message

while True:
   user_message = input("user >>")

   if user_message.lower() == "quit":
     break
   ai_message = chat_with_user(user_message)
   print(f'ai >> {ai_message}')
````
