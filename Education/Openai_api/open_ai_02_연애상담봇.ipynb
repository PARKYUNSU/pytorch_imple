```python
model = 'gpt-3.5-turbo'

query = '''
상대방 MBTI : infp
상대방 이름 : 카리나
관계 : 연인
상황 : 만난 지 100일 된 상황
질문 : 100일 기념 선물로 무엇이 좋을 거 같아? 보통 어떤 것을 선물하니?
'''
# few shot

messages = [
    {'role': 'system', 'content': '연예코치'},
     {'role': 'user', 'content': query}
            ]


try:
    response = openai.ChatCompletion.create(
                            model = model,
                            messages=messages,
                            temperature=0.7
                            # 0에 가까울 수록 일반적 답변
                        )

    answer = response.choices[0].message['content']
    print(answer)

except Exception as e:
    print(f'에러가 발생했어요: {e}')
```

```python
my_mbti = input("내 MBTI : ").upper() # 대문자로 일괄 변환
your_mbti = input("상대방 MBTI : ").upper() # 대문자로 일괄 변환
your_name = input("상대방 이름 : ")
relation = input('관계: ')
situation  = input('상황: ')
question = input('질문: ')

query = f'''
내 MBTI : {my_mbti}
상대방 MBTI : {your_mbti}
상대방 이름 : {your_name}
관계 : {relation}
상황 : {situation}
질문 : {question}
'''

messages = [
    {'role': 'system', 'content': '연예상담사'},
     {'role': 'user', 'content': query}
            ]

print('답변을 준비중입니다...잠시만 기다려 주실거죠')


response = openai.ChatCompletion.create(
                            model = model,
                            messages=messages,
                            temperature=0.7
                            # 0에 가까울 수록 일반적 답변
                        )

answer = response.choices[0].message['content']
messages.append({'role':'assistant', 'content': answer})
print(answer)

# 질문이 끊기지 않고 계속 하도록 while 사용 (multi-turn 구현)

while True:
    # 사용자로부터 질문을 받습니다.
    question = input('질문: ')
    query = f'Q: {question}'
    # 질문을 메시지 배열에 추가합니다.
    messages.append({'role': 'user', 'content': query})
    # chatGPT API 호출 >> 응답 받습니다.
    print('답변을 준비중이예요.....')
    response = openai.ChatCompletion.create(
                            model = model,
                            messages=messages,
                            temperature=0.7
                            # 0에 가까울 수록 일반적 답변
                        )
    answer = response.choices[0].message['content']
    # 응답을 출력합니다.
    print('=='*80)
    print(answer)
    messages.append({'role':'assistant', 'content': answer})
    # 사용자에게 더 질문할 건지 확인
    while True:
        q = input('더 질문하실건가요?(y/n): ').lower() # 입력된 영어문자를 전부 소문자로 변환
        if q in ['y', 'n', 'ㅛ','ㅜ']:
            break  # 정지조건
        else:
            print("유효한 입력이 아닙니다. y 나 n 선택해서 다시 입력해 주세요")

    if q in ['n', 'ㅜ']:
        print("종료합니다.")
        break
```
