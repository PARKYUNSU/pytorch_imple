```python
import openai
import os

from google.colab import userdata
os.environ["OPENAI_API_KEY"] = userdata.get('temp')
```

```python
# open ai model

completion=\
openai.chat.completions.create(
    model = 'gpt-3.5-turbo',
    messages = [
        {"role": "system", "content": "You are a helpful assistant. You must reply in Korean."},
        {"role": "user", "content": "hello~"}
    ]
)

# system, user, assistant, function (주제, 정보 제공자)
# - system : 역할부여 프롬프팅, 지시문, 퓨샷, 지문
# - user : 서비스 시에 사용자가 입력하는 문장
# - assistant : GPT가 생성하는 문장
# - function : 함수나 API 호출 결과
# 위 항목은 "role"이라는 키에 저장합니다.
```

```python
print(completion)
```

```python
completion=\
openai.chat.completions.create(
    model = 'gpt-4o',
    messages = [
        {"role": "system", "content": "You are a helpful assistant. You must write in Korean."},
        {"role": "user", "content": "hello~"}
    ],
    n=3,
    temperature=2.0)

for choice in completion.choices:
    print(choice.message)
    print(choice.message.content)
```
